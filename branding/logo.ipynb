# DeloneCommons — logo

## The name

DeloneCommons is named in honour of **Boris Nikolaevich Delone (1890–1980)**, also known in the mathematical literature as Delaunay, a Russian and Soviet mathematician whose research brought together the geometry of numbers, discrete geometry, and mathematical crystallography.

Although best known today for the Delaunay triangulation, Delone devoted a substantial part of his scientific career to understanding the geometric foundations of crystalline matter. His contributions range from practical methods for crystallographic lattice analysis to fundamental questions about the nature and origin of crystalline order.

### Crystal lattices and their classification

A familiar problem in crystallography is that the same lattice can be described by infinitely many different unit cells. Apparently different cell parameters may therefore describe exactly the same underlying periodic arrangement.

Delone used a lattice reduction method based on the earlier work of Eduard Selling. In three dimensions, the method considers four vectors: the three primitive basis vectors and a fourth vector equal to the negative of their sum. The basis is transformed until the angles between every pair of these four vectors are right or obtuse. The six scalar products between these vectors then provide a compact representation of the reduced lattice.

This provides a systematic geometric representation of the lattice, useful for identifying its symmetry, comparing experimentally determined unit cells, and recognizing equivalent lattice descriptions. Selling–Delone reduction remains relevant to modern computational crystallography, particularly when large collections of experimental unit cells must be compared efficiently.

In 1933, Delone also introduced a classification of three-dimensional lattices into **24 symmetrical sorts**, taking into account both the topology of their Voronoi–Dirichlet cells and their symmetry properties. This is a finer classification than the familiar 14 Bravais lattice types.

For example, the Voronoi cells of primitive cubic, body-centred cubic, and face-centred cubic lattices are a cube, a truncated octahedron, and a rhombic dodecahedron, respectively. Such polyhedra reveal the geometry of the surrounding lattice directly, without relying on a particular choice of unit-cell basis.

These ideas became part of the mathematical apparatus of structural crystallography. In 1934, Delone, N. N. Padurov, and A. D. Alexandrov published a comprehensive monograph on the mathematical foundations of crystal structure analysis and the determination of the fundamental translation cell from X-ray diffraction data.

### The empty-sphere method and Delaunay tessellation

Delone introduced his empty-sphere method at the 1924 International Congress of Mathematicians in Toronto and developed it further in his 1934 paper *Sur la sphère vide*, dedicated to the memory of Georgy Voronoi.

Consider a sphere moving through an arrangement of points, free to change its position and radius, but subject to one condition: no point may lie in its interior.

The convex hull of the points lying on the surface of a maximal empty sphere defines a full-dimensional Delaunay cell. For generic point configurations, these cells are triangles in two dimensions and tetrahedra in three dimensions. Degenerate configurations may give rise to cells with more vertices.

The resulting tessellation is geometrically dual to the Voronoi diagram:

- Voronoi cells describe the regions of space associated with individual points.
- Shared Voronoi faces of codimension one identify neighbouring points and correspond to Delaunay edges.
- Voronoi vertices correspond to the centres of empty circles or spheres defining full-dimensional Delaunay cells.

For crystallography, these complementary representations offer a natural way to investigate coordination environments, interatomic neighbourhoods, packing geometry, and the distribution of empty space within a structure.

### Delone sets: a geometric model of matter

Rather than assuming periodicity from the outset, Delone introduced a more general description of spatial arrangements, originally called $(r,R)$-systems and now known as **Delone sets**.

A Delone set satisfies two simple geometric conditions:

- There is a positive lower bound on the distance between distinct points (at least $2r$, for some $r>0$).
- There is a finite upper bound $R$ on the distance from any location in space to the nearest point.

The first condition represents the idea that material particles cannot accumulate arbitrarily close to one another. The second expresses a broad form of spatial coverage: arbitrarily large empty regions are excluded.

Neither condition requires translational periodicity.

This makes Delone sets a mathematical framework in which periodic crystals, quasiperiodic structures, and certain idealized disordered arrangements can be studied using a common language. Rather than treating crystalline and non-crystalline matter as fundamentally unrelated geometric objects, one can investigate the additional conditions that distinguish different forms of spatial order.

### How local order gives rise to global symmetry

One of the most remarkable questions pursued by Delone and his school was whether the global regularity of an infinite structure could be deduced entirely from its local environments.

Classical crystallography generally starts with a periodic structure and investigates its symmetry. Delone's approach reverses the question: **what local geometric properties are sufficient to guarantee global crystalline order?**

In 1976, Delone, N. P. Dolbilin, M. I. Shtogrin, and R. V. Galiulin established a local criterion for the regularity of a point system.

Consider the cluster of points within a radius $\rho$ of a chosen point in a Delone set. Its local symmetry group consists of the isometries that fix the central point and preserve the cluster. The theorem states that if all clusters of radius $\rho_0+2R$ are equivalent, and the local symmetry group of a cluster has already stabilized between radii $\rho_0$ and $\rho_0+2R$, the system is regular. Every point can then be mapped onto every other point by a symmetry of the entire infinite structure. In fact, the global regular system is determined by one such cluster, within the class of systems with equivalent clusters of that radius.

These conditions are much stronger than simply having identical nearest-neighbour coordination. They provide a rigorous connection between local geometry and global structural organization.

The local theory subsequently developed into the study of structures containing multiple types of local environments, as well as more general questions concerning periodicity, tilings, and spatial order.

For a molecular crystallographer, this perspective is particularly inspiring: it suggests studying crystal packing not only through unit cells and space groups, but also through the geometry, symmetry, and connectivity of local structural environments.

### From mathematical crystallography to crystal chemistry

The geometric tradition represented by Voronoi and Delone has continued to develop within modern crystal chemistry.

Voronoi–Dirichlet partitions have been used to characterize atomic domains, coordination environments, intermolecular contacts, molecular volumes, and cavities in crystalline structures. Their relationship to physically motivated partitions, including those derived from electron density, has also been investigated.

As one example, a 2004 study by I. A. Baburin and V. A. Blatov used molecular Voronoi–Dirichlet polyhedra to examine the volumes of more than 100,000 molecules in organic crystals.

These developments illustrate how abstract geometric constructions can become practical instruments for understanding chemical structure.

DeloneCommons is inspired by this scientific tradition. Its purpose is to develop open computational tools connecting discrete geometry, periodic graphs, symmetry, chemical structure, and molecular-crystal science.

The word *Commons* reflects the intention to make these tools and the underlying knowledge openly accessible and reusable.

## The logo

The DeloneCommons logo is constructed from a genuine two-dimensional Voronoi diagram rather than from a manually drawn letter.

Seven generator points are arranged so that the Voronoi cell of the central point resembles the letter **D**.

The elements of the image represent:

- **Filled and open circles:** the central generator and its neighbouring generators.
- **Bold edges:** the boundary of the central Voronoi cell.
- **Square markers:** the vertices of that cell.
- **Thin solid lines:** the remaining Voronoi edges.
- **Dashed lines:** the Delaunay edges connecting the central generator to its neighbours.

The logo thus combines the spatial partition and its dual neighbourhood graph in a single image, bringing together two complementary ways of describing the same geometric structure.

The following code preserves the original construction, with editable generator coordinates and rendering parameters. SciPy computes the Voronoi geometry, and Matplotlib renders it. The display clips the exterior rays for illustration; it does not introduce extra generating sites or modify the central cell.

````python
import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from scipy.spatial import Voronoi
````

````python
# --- Point input / generation cell ---
# The FIRST point is treated as the central point whose Voronoi cell gets bold edges.
# By default that central point is (0, 0).

central_point = np.array([0.0, 0.0])

# Option A: manually supplied points (central point first)
# This default set makes the central Voronoi cell read like a "D".
manual_points = np.array(
    [
        [0.0, 0.0],
        [-1.25, 0.0],
        [-0.5, 2.5],
        [1.5, 2.0],
        [2.0, 0.0],
        [-0.5, -2.5],
        [1.5, -2.0],
    ],
    dtype=float,
)

# Option B: random ring-like points around the central point
use_random_points = False
rng_seed = 14
rng = np.random.default_rng(rng_seed)

n_random_ring_points = 6
angles = np.sort(rng.uniform(0.0, 2.0 * np.pi, size=n_random_ring_points))
radii = rng.uniform(0.82, 1.08, size=n_random_ring_points)
random_ring = np.column_stack([radii * np.cos(angles), radii * np.sin(angles)])

if use_random_points:
    points = np.vstack([central_point, random_ring])
else:
    points = manual_points.copy()

if not np.allclose(points[0], central_point):
    print("Note: the first point is treated as the central point.")

points
````

````python
def compute_voronoi_data(points, margin):
    '''
    Build a clipped 2D Voronoi diagram with the first point treated as the central point.

    Parameters
    ----------
    points : array-like, shape (n, 2)
        Generator points. The first point is the central point.
    margin : float
        Bounding-box margin. The clipping box is
        [min(x)-margin, max(x)+margin] x [min(y)-margin, max(y)+margin].

    Returns
    -------
    tuple
        (
            points,               # 0: all generator points, central point first
            point_neighbors,      # 1: point-neighbor lists (indices into `points`)
            nodes,                # 2: Voronoi node coordinates after clipping
            central_cell_nodes,   # 3: node indices forming the central cell
            central_cell_edges,   # 4: node-index pairs forming the central cell
            all_edges,            # 5: all clipped Voronoi edges (node-index pairs)
            bbox,                 # 6: (xmin, xmax, ymin, ymax)
        )
    '''
    points = np.asarray(points, dtype=float)

    if points.ndim != 2 or points.shape[1] != 2:
        raise ValueError("points must be an (n, 2) array.")
    if len(points) < 2:
        raise ValueError("Need at least two points.")
    if margin <= 0:
        raise ValueError("margin must be positive.")

    rounded_points = np.round(points, 12)
    if len(np.unique(rounded_points, axis=0)) != len(points):
        raise ValueError("Duplicate generator points are not allowed.")

    vor = Voronoi(points)

    xmin = float(points[:, 0].min() - margin)
    xmax = float(points[:, 0].max() + margin)
    ymin = float(points[:, 1].min() - margin)
    ymax = float(points[:, 1].max() + margin)
    bbox = (xmin, xmax, ymin, ymax)

    tol = 1e-10
    site_center = points.mean(axis=0)

    point_neighbors = [set() for _ in range(len(points))]
    for i, j in vor.ridge_points:
        i = int(i)
        j = int(j)
        point_neighbors[i].add(j)
        point_neighbors[j].add(i)

    nodes = []
    node_lookup = {}
    all_edges = set()
    central_edges = set()

    def add_node(coord):
        coord = np.asarray(coord, dtype=float)
        key = tuple(np.round(coord, 10))
        if key not in node_lookup:
            node_lookup[key] = len(nodes)
            nodes.append(coord)
        return node_lookup[key]

    def clip_segment_to_box(p0, p1):
        x0, y0 = map(float, p0)
        x1, y1 = map(float, p1)
        dx = x1 - x0
        dy = y1 - y0

        # Liang–Barsky clipping against the axis-aligned box.
        p = (-dx, dx, -dy, dy)
        q = (x0 - xmin, xmax - x0, y0 - ymin, ymax - y0)
        u1, u2 = 0.0, 1.0

        for pi, qi in zip(p, q):
            if abs(pi) < tol:
                if qi < 0:
                    return None
            else:
                r = qi / pi
                if pi < 0:
                    if r > u2:
                        return None
                    if r > u1:
                        u1 = r
                else:
                    if r < u1:
                        return None
                    if r < u2:
                        u2 = r

        start = np.array([x0 + u1 * dx, y0 + u1 * dy], dtype=float)
        end = np.array([x0 + u2 * dx, y0 + u2 * dy], dtype=float)
        return start, end

    def ray_box_intersection(origin, direction):
        origin = np.asarray(origin, dtype=float)
        direction = np.asarray(direction, dtype=float)
        candidates = []

        if abs(direction[0]) > tol:
            for x in (xmin, xmax):
                t = (x - origin[0]) / direction[0]
                if t > tol:
                    y = origin[1] + t * direction[1]
                    if ymin - tol <= y <= ymax + tol:
                        candidates.append((t, np.array([x, np.clip(y, ymin, ymax)], dtype=float)))

        if abs(direction[1]) > tol:
            for y in (ymin, ymax):
                t = (y - origin[1]) / direction[1]
                if t > tol:
                    x = origin[0] + t * direction[0]
                    if xmin - tol <= x <= xmax + tol:
                        candidates.append((t, np.array([np.clip(x, xmin, xmax), y], dtype=float)))

        if not candidates:
            return None

        candidates.sort(key=lambda item: item[0])
        return candidates[0][1]

    for (site_i, site_j), ridge_vertices in zip(vor.ridge_points, vor.ridge_vertices):
        site_i = int(site_i)
        site_j = int(site_j)

        finite_vertices = [v for v in ridge_vertices if v >= 0]

        if len(finite_vertices) == 2:
            start = vor.vertices[finite_vertices[0]]
            end = vor.vertices[finite_vertices[1]]
        elif len(finite_vertices) == 1:
            start = vor.vertices[finite_vertices[0]]

            tangent = points[site_j] - points[site_i]
            tangent = tangent / np.linalg.norm(tangent)
            normal = np.array([-tangent[1], tangent[0]], dtype=float)

            midpoint = 0.5 * (points[site_i] + points[site_j])
            sign = np.sign(np.dot(midpoint - site_center, normal))
            if sign == 0:
                sign = 1.0
            direction = sign * normal

            end = ray_box_intersection(start, direction)
            if end is None:
                continue
        else:
            continue

        clipped = clip_segment_to_box(start, end)
        if clipped is None:
            continue

        clipped_start, clipped_end = clipped
        if np.linalg.norm(clipped_end - clipped_start) < 1e-8:
            continue

        node_i = add_node(clipped_start)
        node_j = add_node(clipped_end)

        edge = tuple(sorted((node_i, node_j)))
        all_edges.add(edge)

        if site_i == 0 or site_j == 0:
            central_edges.add(edge)

    if nodes:
        nodes = np.vstack(nodes)
    else:
        nodes = np.zeros((0, 2), dtype=float)

    central_node_set = sorted({idx for edge in central_edges for idx in edge})
    if central_node_set:
        angles = [
            math.atan2(nodes[idx, 1] - points[0, 1], nodes[idx, 0] - points[0, 0])
            for idx in central_node_set
        ]
        central_cell_nodes = [idx for _, idx in sorted(zip(angles, central_node_set))]
    else:
        central_cell_nodes = []

    point_neighbors = [sorted(neigh) for neigh in point_neighbors]
    central_cell_edges = sorted(central_edges)
    all_edges = sorted(all_edges)

    return (
        points,
        point_neighbors,
        nodes,
        central_cell_nodes,
        central_cell_edges,
        all_edges,
        bbox,
    )
````

````python
def plot_voronoi_data(
    voronoi_data,
    ax=None,
    point_styles=None,
    edge_styles=None,
    node_styles=None,
    line_styles=None,
    show_box=False,
    square_view=True,
    view_source="points",
    view_pad=0.02,
):
    '''
    Plot the clipped Voronoi geometry with separately stylable point / edge / node / line groups.

    Parameters
    ----------
    voronoi_data : tuple
        Output of `compute_voronoi_data(...)`.
    ax : matplotlib.axes.Axes or None
        Existing axes. If None, a new square figure is created.
    point_styles, edge_styles, node_styles, line_styles : dict or None
        Optional style overrides. Supported keys:

        point_styles:
            "central", "neighbor", "other"
        edge_styles:
            "central", "other"
        node_styles:
            "central", "other"
        line_styles:
            "central_to_neighbor", "neighbor_to_neighbor", "other"

        Important behavior:
        - for line_styles, an empty dict `{}` means "do not draw this family"
        - for point / edge / node styles, an empty dict `{}` keeps the shared defaults
    show_box : bool
        If True, draw the clipping box.
    square_view : bool
        If True, choose x/y limits with equal span for a square composition.
    view_source : {"points", "points+central_nodes", "points+all_nodes", "bbox"}
        Geometry used to choose plot limits when `square_view=True`.
        "points" usually gives the cleanest logo framing.
    view_pad : float
        Fractional padding applied to the larger x/y span when `square_view=True`.
    '''
    (
        points,
        point_neighbors,
        nodes,
        central_cell_nodes,
        central_cell_edges,
        all_edges,
        bbox,
    ) = voronoi_data[:7]

    xmin, xmax, ymin, ymax = bbox

    # Shared defaults: if no custom styles are supplied, the result is a plain Voronoi image.
    default_point_base = dict(
        s=32,
        color="black",
        marker="o",
        zorder=8,
    )
    default_edge_base = dict(
        color="black",
        linewidth=1.15,
        solid_capstyle="butt",
        zorder=3,
    )
    default_node_base = dict(
        s=18,
        color="black",
        marker="s",
        zorder=8,
    )

    default_point_styles = {
        "central": default_point_base.copy(),
        "neighbor": default_point_base.copy(),
        "other": default_point_base.copy(),
    }
    default_edge_styles = {
        "central": default_edge_base.copy(),
        "other": default_edge_base.copy(),
    }
    default_node_styles = {
        "central": default_node_base.copy(),
        "other": default_node_base.copy(),
    }
    default_line_styles = {
        "central_to_neighbor": {},
        "neighbor_to_neighbor": {},
        "other": {},
    }

    def merge_styles(defaults, overrides):
        merged = {key: value.copy() for key, value in defaults.items()}
        if overrides is not None:
            for key, value in overrides.items():
                merged.setdefault(key, {})
                if value:
                    merged[key].update(value)
        return merged

    def merge_line_styles(defaults, overrides):
        merged = {key: value.copy() for key, value in defaults.items()}
        if overrides is not None:
            for key, value in overrides.items():
                if value is None or value == {}:
                    merged[key] = {}
                else:
                    merged.setdefault(key, {})
                    merged[key].update(value)
        return merged

    point_styles = merge_styles(default_point_styles, point_styles)
    edge_styles = merge_styles(default_edge_styles, edge_styles)
    node_styles = merge_styles(default_node_styles, node_styles)
    line_styles = merge_line_styles(default_line_styles, line_styles)

    if ax is None:
        fig, ax = plt.subplots(figsize=(6.5, 6.5), facecolor="white")
        fig.subplots_adjust(left=0.0, right=1.0, bottom=0.0, top=1.0)
    else:
        fig = ax.figure

    central_neighbors = set(point_neighbors[0])

    point_groups = {
        "central": [0],
        "neighbor": sorted(central_neighbors),
        "other": [idx for idx in range(len(points)) if idx != 0 and idx not in central_neighbors],
    }

    central_edge_set = {tuple(sorted(edge)) for edge in central_cell_edges}
    other_edges = [edge for edge in all_edges if tuple(sorted(edge)) not in central_edge_set]

    central_node_set = set(central_cell_nodes)
    node_groups = {
        "central": central_cell_nodes,
        "other": [idx for idx in range(len(nodes)) if idx not in central_node_set],
    }

    all_point_pairs = sorted(
        {
            tuple(sorted((i, j)))
            for i, neigh_list in enumerate(point_neighbors)
            for j in neigh_list
            if i < j
        }
    )

    central_to_neighbor_lines = [(0, j) for j in sorted(central_neighbors)]
    neighbor_to_neighbor_lines = [
        pair
        for pair in all_point_pairs
        if pair[0] in central_neighbors and pair[1] in central_neighbors
    ]

    taken = {
        tuple(sorted(pair))
        for pair in (central_to_neighbor_lines + neighbor_to_neighbor_lines)
    }
    other_lines = [
        pair for pair in all_point_pairs if tuple(sorted(pair)) not in taken
    ]

    def draw_point_lines(pairs, style):
        for i, j in pairs:
            ax.plot(
                [points[i, 0], points[j, 0]],
                [points[i, 1], points[j, 1]],
                **style,
            )

    def draw_voronoi_edges(edge_list, style):
        for node_i, node_j in edge_list:
            ax.plot(
                [nodes[node_i, 0], nodes[node_j, 0]],
                [nodes[node_i, 1], nodes[node_j, 1]],
                **style,
            )

    for line_style_name, lines in (
        ("other", other_lines),
        ("neighbor_to_neighbor", neighbor_to_neighbor_lines),
        ("central_to_neighbor", central_to_neighbor_lines),
    ):
        line_style_dict = line_styles.get(line_style_name, {})
        if line_style_dict:
            draw_point_lines(lines, line_style_dict)

    for edge_style_name, edges in (
        ("other", other_edges),
        ("central", central_cell_edges),
    ):
        edge_style_dict = edge_styles.get(edge_style_name, {})
        if edge_style_dict:
            draw_voronoi_edges(edges, edge_style_dict)

    for group_name in ("other", "central"):
        idxs = node_groups[group_name]
        if idxs:
            ax.scatter(nodes[idxs, 0], nodes[idxs, 1], **node_styles[group_name])

    for group_name in ("other", "neighbor", "central"):
        idxs = point_groups[group_name]
        if idxs:
            ax.scatter(points[idxs, 0], points[idxs, 1], **point_styles[group_name])

    if show_box:
        box_x = [xmin, xmax, xmax, xmin, xmin]
        box_y = [ymin, ymin, ymax, ymax, ymin]
        ax.plot(box_x, box_y, color="0.75", linewidth=0.9, linestyle="--", zorder=0)

    if square_view:
        if view_source == "points":
            view_coords = points
        elif view_source == "points+central_nodes":
            if central_cell_nodes:
                view_coords = np.vstack([points, nodes[central_cell_nodes]])
            else:
                view_coords = points
        elif view_source == "points+all_nodes":
            if len(nodes):
                view_coords = np.vstack([points, nodes])
            else:
                view_coords = points
        elif view_source == "bbox":
            view_coords = np.array([[xmin, ymin], [xmax, ymax]], dtype=float)
        else:
            raise ValueError(
                "view_source must be one of: "
                "'points', 'points+central_nodes', 'points+all_nodes', 'bbox'."
            )

        vxmin = float(view_coords[:, 0].min())
        vxmax = float(view_coords[:, 0].max())
        vymin = float(view_coords[:, 1].min())
        vymax = float(view_coords[:, 1].max())

        cx = 0.5 * (vxmin + vxmax)
        cy = 0.5 * (vymin + vymax)
        span = max(vxmax - vxmin, vymax - vymin)
        half_span = 0.5 * span + view_pad * span

        ax.set_xlim(cx - half_span, cx + half_span)
        ax.set_ylim(cy - half_span, cy + half_span)
    else:
        ax.set_xlim(xmin, xmax)
        ax.set_ylim(ymin, ymax)

    ax.set_aspect("equal", adjustable="box")
    try:
        ax.set_box_aspect(1)
    except AttributeError:
        pass

    ax.axis("off")
    return fig, ax
````

````python
margin = 0.60
voronoi_data = compute_voronoi_data(points, margin=margin)

(
    points_out,
    point_neighbors,
    nodes,
    central_cell_nodes,
    central_cell_edges,
    all_edges,
    bbox,
) = voronoi_data

print(f"Number of generator points: {len(points_out)}")
print(f"Number of Voronoi nodes: {len(nodes)}")
print(f"Central-point neighbors: {point_neighbors[0]}")
print(f"Central-cell node indices: {central_cell_nodes}")
print(f"Central-cell edges: {central_cell_edges}")
print(f"Bounding box: {bbox}")
print("Recommended square view: view_source='points', view_pad=0.02")
````
